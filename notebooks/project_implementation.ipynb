## 1. Original ML objective
Predict India's current cumulative confirmed COVID-19 cases using the previous day's confirmed cases.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/full_grouped.csv')
india = df[df['Country/Region'] == 'India'].copy()
india['Date'] = pd.to_datetime(india['Date'])
india = india.sort_values('Date')
print('Dataset:', df.shape)
print('India:', india.shape)
india.head()
```
